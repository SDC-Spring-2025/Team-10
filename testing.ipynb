```python
list_totals = [56.58, 69.25, 7.61, 5.35, 15.03, 179.94, 93.58, 143.71, 24.47, 30.76, 77.83, 64.43, 33.26, 9.58, 48.53, 50.29, 35.52, 11.04, 2.17, 45.58, 47.02, 69.76, 37.37, 30.15, 33.92, 7.27, 51.82, 41.31, 34.21, 30.98, 94.78, 44.35, 9.50, 46.73, 17.00, 26.38, 52.76, 164.89, 17.80, 13.05, 28.04, 17.00, 55.39, 26.67, 10.94, 22.03, 38.86, 85.07, 91.45, 77.49]
```

```python
def compare_to(image_number):
  if float(process_receipt(config, "backend/receipts/10{}-receipt.jpg".format(image_number), out_dir=None, verbosity=0).sum) == list_totals[int(image_number)]:
    return True
  else:
    return False

def find_total(image_number):
  return process_receipt(config, "backend/receipts/10{}-receipt.jpg".format(image_number), out_dir=None, verbosity=0).sum
```

```python
print(find_total("01"))
```

```python
def test_amount():
  true_false_list = []
  for i in range(50):
    if i < 10:
      true_false_list.append(compare_to("0{}".format(i)))
    else:
      true_false_list.append(compare_to(i))
  return true_false_list

def find_all_totals():
  list_of_totals = []
  for i in range(50):
    print(i)
    if i < 10:
      list_of_totals.append(find_total("0{}".format(i)))
    else:
      list_of_totals.append(find_total(i))
  return list_of_totals
```

```python
true_false_list = test_amount()
```
